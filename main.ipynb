# Обработка текста

## Русский текст

Уберём все знаки препинания, цифры, ненужные пробелы, пропуски и приведём текст к нижнему регистру. Римскую нумерацию глав переведём в арабскую.

```python
import re


def Russian_preprocessing(file_path):
    with open(file_path, 'r', encoding='windows-1251') as file:
        text = file.read()
    
    # Разбиваем на строки для сохранения структуры стихотворения
    lines = text.split('\n')
    romanNums = set('IVXLCDMХ')
    processed_lines = []
    chapterPattern = r'^\s*Глава\s+[а-яё]+\n*$'
    cur_poem = ''
    curPoemIndex = 0
    for line in lines:
        if line and set(line) <= romanNums:
            if cur_poem:
                curPoemIndex += 1
                cur_poem = str(curPoemIndex) + ":"+ cur_poem
                processed_lines.append(cur_poem)
            cur_poem = ''
        else:
            line = re.sub(r'Х','', line)
            line = re.sub(chapterPattern,'', line)
            # Убираем числа
            line = re.sub(r'\d+', '', line)
            
            # Убираем знаки препинания, но сохраняем дефисы
            line = re.sub(r'[^\w\s]', ' ', line)
            
            # Оставляем только русские буквы и пробелы
            line = re.sub(r'[^а-яёА-ЯЁ\s]', '', line)
            
            # Убираем лишние пробелы
            line = re.sub(r'\s+', ' ', line).strip()
            line = line.lower()
            cur_poem += (line + ' ')
    if cur_poem:
        processed_lines.append(str(curPoemIndex + 1) + ':'+cur_poem)
    return '\n'.join(processed_lines)

# Использование
file_path = 'data\PushkinRuRow.txt'
processed_text = Russian_preprocessing(file_path)

# Сохраняем результат
with open('data\PushkinRuProcessed.txt', 'w', encoding='windows-1251') as file:
    file.write(processed_text)

print(f"Обработка завершена! Сохранено {len(processed_text.splitlines())} строк.")
```

## Английский текст

```python
import re

def English_preprocessing(file_path):
    with open(file_path, 'r', encoding='windows-1251') as file:
        text = file.read()
    
    # Разбиваем на строки для сохранения структуры стихотворения
    lines = text.split('\n')
    romanNums = set('IVXLCDM')
    processed_lines = []
    chapterPattern = r'^\s*Chapter\s+[A-Za-z]+\n*$'
    cur_poem = ''
    curPoemIndex = 0
    for line in lines:
        line = line.strip()
        if line and set(line) <= romanNums:
            curPoemIndex += 1
            if cur_poem:
                processed_lines.append(cur_poem)
            cur_poem =  str(curPoemIndex) + ':'
        else:
            line = re.sub(r'Х','', line)
            line = re.sub(chapterPattern,'', line)
            # Убираем числа
            line = re.sub(r'\d+', '', line)
            
            # Убираем знаки препинания, но сохраняем дефисы
            line = re.sub(r'[^\w\s]', ' ', line)
            
            # Оставляем только русские буквы и пробелы
            line = re.sub(r'[^a-zA-Z\s]', '', line)
            
            # Убираем лишние пробелы
            line = re.sub(r'\s+', ' ', line).strip()
            line = line.lower()
            cur_poem += (line + ' ')
    if cur_poem:
        processed_lines.append(cur_poem)
    return '\n'.join(processed_lines)

# Использование
file_path = 'data\PushkinEngRow.txt'
processed_text = English_preprocessing(file_path)

# Сохраняем результат
with open('data\PushkinENProcessed.txt', 'w', encoding='windows-1251') as file:
    file.write(processed_text)

print(f"Обработка завершена! Сохранено {len(processed_text.splitlines())} строк.")
```
